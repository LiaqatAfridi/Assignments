### Q5
`numbers.txt` file ko read karein aur tamam numbers ka sum (total) calculate karke print karein.

```python
# Q5 solution code here
file = open("numbers.txt", "r")
lines = file.readlines()
file.close()

total = 0
for line in lines:
    total += int(line.strip())

print("Sum of numbers:", total)
```

### Q7 (Medium)

### Q9 
Ek file `source.txt` ka data read karke `destination.txt` file mein copy karein.

```python
# Q9 solution code here
# Creating source.txt first
with open("source.txt", "w") as file:
    file.write("This data will be copied to destination.txt")

# copy
with open("source.txt", "r") as source_file:
    data = source_file.read()

# paste
with open("destination.txt", "w") as dest_file:
    dest_file.write(data)

print("Data copied successfully.")
```

### Q17
Ek `try-except` block likhein jo Multiple Exceptions catch kar sake (`ZeroDivisionError` aur `ValueError`).

```python
# Q17 solution code here
try:
    num = int(input("Enter a number to divide 10 by: "))
    result = 10 / num
    print(result)
except ZeroDivisionError:
    print("Error: Cannot divide by zero.")
except ValueError:
    print("Error: Please enter a valid number.")
```

### Q18
`else` block ka istemal `try-except` ke sath karein jo sirf tab chale jab koi error na aaye.

```python
# Q18 solution code here
try:
    price = float(input("Enter product price: "))
    quantity = int(input("Enter quantity: "))
    total = price * quantity

except ValueError:
    print("Error: Please enter a valid number.")

else:
    print("Total price is:", total)
```

### Q19 
`raise` keyword ka use karke age validation ka function banayein. Agar age 18 se kam ho toh custom `ValueError("Age must be 18 or older")` raise karein.

```python
# Q19 solution code here
def check_age(age):
    if age < 18:
        raise ValueError("Age must be 18 or older")
    else:
        print("Age is valid:", age)

try:
    check_age(15)
except ValueError as e:
    print("Error:", e)
```

### Q20 
File handling aur Exception handling ko combine karein: Ek file read karein, agar file na mile toh error catch karein, aur `finally` block mein file closure handle karein.

```python
# Q20 solution code here
file_handle = None

try:
    file_handle = open("does_not_exist.txt", "r")
    content = file_handle.read()
    print(content)
except FileNotFoundError:
    print("Error: File not found.")
finally:
    if file_handle is not None:
        file_handle.close()
        print("File closed successfully.")
    else:
        print("File was never opened, so nothing to close.")
```
