### **Question 36**
(Visualization) 'Drama' aur 'Action' genres ki movies ke 'Gross' revenue ko compare karne ke liye ek Grouped Box Plot ya Violin Plot banayein.

```python
# A movie can have both genres (e.g. "Action, Drama"), so it will appear in both groups
drama = data[data.Genre.str.contains("Drama")].copy()
drama["Genre_Group"] = "Drama"

action = data[data.Genre.str.contains("Action")].copy()
action["Genre_Group"] = "Action"

compare = pd.concat([drama, action])
compare["Gross_Million"] = compare.Gross / 1_000_000

plt.figure(figsize=(8, 6))
sns.boxplot(x="Genre_Group", y="Gross_Million", data=compare,
            hue="Genre_Group", palette=["skyblue", "salmon"], legend=False)
plt.title("Gross Revenue: Drama vs Action")
plt.xlabel("Genre")
plt.ylabel("Gross (in Million $)")
plt.show()
```

### **Question 40**
(Visualization) Top 5 highest rated movies aur unke 'Meta_score' vs 'IMDB_Rating' (scaled to 100) ka ek Side-by-Side Comparison Bar Chart banayein.

```python
import numpy as np

top5 = data.nlargest(5, "IMDB_Rating")
x = np.arange(len(top5))
width = 0.35

plt.figure(figsize=(10, 5))
plt.bar(x - width / 2, top5.IMDB_Rating * 10, width, label="IMDB Rating (x10)", color="royalblue")
plt.bar(x + width / 2, top5.Meta_score, width, label="Meta Score", color="orange")
plt.xticks(x, top5.Series_Title, rotation=20, ha="right")
plt.title("Top 5 Highest Rated Movies: IMDB Rating vs Meta Score")
plt.ylabel("Score (out of 100)")
plt.ylim(0, 115)
plt.legend(loc="upper right")
plt.tight_layout()
plt.show()
```
