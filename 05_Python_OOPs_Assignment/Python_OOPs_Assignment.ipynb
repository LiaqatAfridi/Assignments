**Q1:** Aik `Student` class banayein jismein `name` aur `roll_number` private variables (`__`) hon. In variables ko access aur modify karne ke liye getter aur setter methods likhein.

```python
class Student:
    def __init__(self, name, roll_number):
        self.__name = name
        self.__roll_number = roll_number

    def get_name(self):
        return self.__name

    def set_name(self, name):
        self.__name = name

    def get_roll_number(self):
        return self.__roll_number

    def set_roll_number(self, roll_number):
        self.__roll_number = roll_number

s1 = Student("Ali", 101)
print("Name:", s1.get_name())
print("Roll Number:", s1.get_roll_number())

s1.set_name("Ahmed")
s1.set_roll_number(202)
print("Updated Name:", s1.get_name())
print("Updated Roll Number:", s1.get_roll_number())
```

**Q2:** Aik `BankAccount` class banayein jismein `__balance` private variable ho. Aik `deposit()` aur aik `withdraw()` method banayein, check karein ke withdraw karte waqt balance zero se kam na ho.

```python
class BankAccount:
    def __init__(self, balance):
        self.__balance = balance

    def deposit(self, amount):
        if amount > 0:
            self.__balance += amount
            print("Deposited:", amount)
        else:
            print("Deposit amount must be greater than 0")

    def withdraw(self, amount):
        if amount <= self.__balance:
            self.__balance -= amount
            print("Withdrawn:", amount)
        else:
            print("Insufficient balance")

    def get_balance(self):
        return self.__balance

account = BankAccount(1000)
account.deposit(500)
account.withdraw(300)
account.withdraw(5000)
print("Current Balance:", account.get_balance())
```

**Q3:** Aik class banayein `Employee` jismein salary private ho. Agar koi salary ko minus (negative) value set karne ki koshish kare, toh setter method 'Invalid Salary' ka error message show kare.

```python
class Employee:
    def __init__(self, name, salary):
        self.name = name
        self.__salary = salary

    def get_salary(self):
        return self.__salary

    def set_salary(self, salary):
        if salary < 0:
            print("Invalid Salary")
        else:
            self.__salary = salary

emp = Employee("Sara", 50000)
print("Salary:", emp.get_salary())

emp.set_salary(-100)
print("Salary after invalid try:", emp.get_salary())

emp.set_salary(60000)
print("Salary after valid update:", emp.get_salary())
```

**Q4:** Protected variable (`_`) aur Private variable (`__`) ke beech kya farq hota hai? Aik simple code ke zariye dono ka use-case dikhayein.

```python
class Demo:
    def __init__(self):
        self._protected = "I am protected"
        self.__private = "I am private"

class Child(Demo):
    def show(self):
        print(self._protected)       # protected child class mn use hoga

obj = Demo()
print(obj._protected)                # works, laiken outside use nahi karskte

try:
    print(obj.__private)             # does not work
except AttributeError as e:
    print("Error:", e)

child = Child()
child.show()

# Difference:
# _protected  -> just a naming convention, still accessible hota h, meant for child classes.
# __private   -> name mangling happens, outside se directly access nahi hosakta
```

**Q5:** Encapsulation ka real-world faida kya hai? Private variable ko class ke bahar se direct access karne par kaunsa error aata hai? Code likh kar check karein.

```python
class Account:
    def __init__(self):
        self.__pin = 1234

    def check_pin(self, pin):
        return pin == self.__pin
        
acc = Account()
print(acc.check_pin(1234))

try:
    print(acc.__pin)
except AttributeError as e:
    print("AttributeError:", e)
# Benefit of encapsulation: sensitive data (like a pin) is hidden and can only
# be used through methods, so nobody can change or see it directly by mistake.
```

**Q9:** Child class ke constructor (`__init__`) ke andar parent class ke constructor ko call karne ke liye `super()` ka use kaise hota hai? Aik practical example likhein.

```python
class Person:
    def __init__(self, name):
        self.name = name

class Student(Person):
    def __init__(self, name, subject):
        super().__init__(name)      # calling parent constructor
        self.subject = subject

t = Student("Liaqat Khan", "Python")
print("Name:", t.name)
print("Subject:", t.subject)
```

**Q10:** Hierarchical Inheritance kya hoti hai? Aik parent class `Shape` banayein aur us se do alag child classes `Circle` aur `Square` banakar samjhayein.

```python
# Hierarchical inheritance means one parent class has more than one child class.
class Shape:
    def __init__(self, name):
        self.name = name

    def show(self):
        print("This is a", self.name)

class Circle(Shape):
    def __init__(self, radius):
        super().__init__("Circle")
        self.radius = radius

    def area(self):
        return 3.14 * self.radius * self.radius

class Square(Shape):
    def __init__(self, side):
        super().__init__("Square")
        self.side = side

    def area(self):
        return self.side * self.side

c = Circle(5)
s = Square(4)

c.show()
print("Area:", c.area())
s.show()
print("Area:", s.area())
```

**Q11:** Method Overriding kya hai? Parent class `Bird` banayein jisme `fly()` method ho jo print kare 'Most birds can fly'. Child class `Ostrich` banayein jo is method ko override kare aur print kare 'Ostriches cannot fly'.

```python
# Method overriding: child class gives its own version of the parent's method.
class Bird:
    def fly(self):
        print("Most birds can fly")

class Ostrich(Bird):
    def fly(self):
        print("Ostriches cannot fly")

b = Bird()
o = Ostrich()

b.fly()
o.fly()
```

**Q12:** Polymorphism function ke zariye kaise achieve hoti hai? Aik function banayein `make_sound(animal_object)` jo har tarah ke animal class ka sound method automatically call kare.

```python
# Polymorphism: same function works with different objects.
class Dog:
    def sound(self):
        print("Woof Woof")

class Cat:
    def sound(self):
        print("Meow Meow")

class Cow:
    def sound(self):
        print("Moo Moo")

def make_sound(animal_object):
    animal_object.sound()

make_sound(Dog())
make_sound(Cat())
make_sound(Cow())
```

**Q13:** Operator Overloading kya hoti hai? Do objects ko `+` operator ke zariye add karne ke liye `__add__` magic method ka use kaise karte hain? Code likhein.

```python
# Operator overloading means giving a special meaning to an operator for our own class.
class Number:
    def __init__(self, value):
        self.value = value

    def __add__(self, other):
        return Number(self.value + other.value)

n1 = Number(10)
n2 = Number(20)
n3 = n1 + n2
print("Result:", n3.value)
```

**Q16:** `abc` module ka use karte hue aik abstract class `Computer` banayein jisme abstract method `process()` ho. Is class ka object banane ki koshish karein aur dekhein kya hota hai.

```python
from abc import ABC, abstractmethod

class Computer(ABC):
    @abstractmethod
    def process(self):
        pass

try:
    pc = Computer()
except TypeError as e:
    print("Error:", e)
```

**Q18:** Aik abstract class `Car` banayein jisme aik concrete (normal) method `start_engine()` ho aur aik abstract method `drive()` ho. Child class bana kar dono ko test karein.

```python
from abc import ABC, abstractmethod

class Car(ABC):
    def start_engine(self):
        print("Engine started")

    @abstractmethod
    def drive(self):
        pass

class Run(Car):
    def drive(self):
        print("driving smoothly")

my_car = Run()
my_car.start_engine()
my_car.drive()
```

**Q20:** Aik real-world scenario (jaise Payment System) design karein jahan `Payment` aik abstract class ho aur `CreditCardPayment` aur `UPIPayment` uski subclasses hon.

```python
from abc import ABC, abstractmethod

class Payment(ABC):
    @abstractmethod
    def pay(self, amount):
        pass

class CreditCardPayment(Payment):
    def pay(self, amount):
        print("Paid", amount, "using Credit Card")

class UPIPayment(Payment):
    def pay(self, amount):
        print("Paid", amount, "using UPI")

p1 = CreditCardPayment()
p2 = UPIPayment()
p1.pay(5000)
p2.pay(2500)

# Credit card is not the best option because it charges interest 
# if the full amount is not paid by the due date. Muslims should be aware
# that interest (riba) is haram (prohibited) in Islam.
```
