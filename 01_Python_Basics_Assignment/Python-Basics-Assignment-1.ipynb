# Python Basics Assignment
**Topics:** print(), Variables, Data Types, Arithmetic Operators, Logical Operators

## 1. Student Information

```python
# Student information variables
name = "Liaqat Khan"
age = 26
city = "Karachi"
is_student = True

print("Student Name:", name)
print("Age:", age)
print("City:", city)
print("Is Student:", is_student)
```

## 2. Data Types

```python
# One variable of each basic data type
string_val = "AssalamuAlaikum"
integer_val = 45
float_val = 12.75
bool_val = False

print(f"{(string_val):<20} | Type: {type(string_val)}")
print(f"{(integer_val):<20} | Type: {type(integer_val)}")
print(f"{(float_val):<20} | Type: {type(float_val)}")
print(f"{str(bool_val):<20} | Type: {type(bool_val)}")
```

## 3. Arithmetic Operations

```python
# Two numbers for arithmetic operations
num1 = 18
num2 = 4

addition = num1 + num2
subtraction = num1 - num2
multiplication = num1 * num2
division = num1 / num2
modulus = num1 % num2
power = num1 ** num2

print("Addition:", addition)
print("Subtraction:", subtraction)
print("Multiplication:", multiplication)
print("Division:", division)
print("Modulus:", modulus)
print("Power:", power)
```

## 4. Student Marks Calculator

```python
# Marks of three subjects
biochem_marks = 85
biostat_marks = 74
math_marks = 90

total_marks = biochem_marks + biostat_marks + math_marks
average_marks = total_marks / 3

print("Total Marks:", total_marks)
print("Average Marks:", average_marks)
```

## 5. Shopping Bill

```python
# Product price and quantity
price = 250
quantity = 6

total_bill = price * quantity
print("Total Bill:", total_bill)
```

## 6. Rectangle Area

```python
# area and perimeter of a rectangle
length = 12
width = 5

area = length * width
perimeter = 2 * (length + width)

print("Area:", area)
print("Perimeter:", perimeter)
```

## 7. Time Converter

```python
# Convert minutes into hours and remaining minutes
total_mins = 250

hours = total_mins // 60
remaining_mins = total_mins % 60

print("Hours:", hours)
print("Remaining Minutes:", remaining_mins)
```

## 8. Student Eligibility

```python
# Marks and attendance for eligibility check
marks = 79
attendance = 80

is_eligible = marks >= 50 and attendance >= 75
print("Is Eligible:", is_eligible)
```

## 9. Logical Operators

```python
# Boolean variables and logical operators
a = True
b = False

print("a and b:", a and b)
print("a or b:", a or b)
print("not a:", not a)
print("not b:", not b)
```

## Final Challenge: Student Report
**Restriction:** Do not use `if`, `else`, or `elif`

```python
# Student details
student_name = "Liaqat Khan"
student_age = 23
subject1_marks = 70
subject2_marks = 65
subject3_marks = 80
attendance_percentage = 80

# 1. Student information
print("Student Name:", student_name)
print("Age:", student_age)

# 2. Total marks
total = subject1_marks + subject2_marks + subject3_marks
print("Total Marks:", total)

# 3. Average marks
average = total / 3
print("Average Marks:", average)

# 4. Eligibility using logical operators
eligible = average >= 50 and attendance_percentage >= 75
print("Is Eligible:", eligible)
```
